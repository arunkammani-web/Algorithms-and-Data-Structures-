```python
# Bubble Sort Algorithm

def bubble_sort(data):

    # Get total number of elements in the list
    n = len(data)

    # Outer loop for number of passes
    for i in range(n):

        # Variable to track swapping
        swapped = False

        # Compare adjacent elements
        for j in range(0, n - i - 1):

            # Swap if current element is greater
            if data[j] > data[j + 1]:

                # Swapping values
                data[j], data[j + 1] = data[j + 1], data[j]

                swapped = True

        # Stop loop if list is already sorted
        if not swapped:
            break

    return data


# Example dataset
numbers = [45, 12, 78, 3, 19]

# Display sorted output
print("Bubble Sort Result:")
print(bubble_sort(numbers))
```

```python
# Selection Sort Algorithm

def selection_sort(data):

    # Traverse entire list
    for i in range(len(data)):

        # Assume first unsorted element is minimum
        min_index = i

        # Search for smaller element
        for j in range(i + 1, len(data)):

            # Update minimum index if smaller value found
            if data[j] < data[min_index]:
                min_index = j

        # Swap smallest element with current position
        data[i], data[min_index] = data[min_index], data[i]

    return data


# Example dataset
numbers = [64, 25, 12, 22, 11]

# Display sorted output
print("Selection Sort Result:")
print(selection_sort(numbers))
```

```python
# Merge Sort Algorithm

def merge_sort(data):

    # Base condition
    if len(data) <= 1:
        return data

    # Find middle position
    mid = len(data) // 2

    # Divide list into two halves
    left_half = data[:mid]
    right_half = data[mid:]

    # Recursively sort both halves
    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    # Merge sorted halves
    return merge(left_half, right_half)


# Function to merge two sorted lists
def merge(left, right):

    # Empty list to store merged values
    sorted_list = []

    i = 0
    j = 0

    # Compare elements from both lists
    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            sorted_list.append(left[i])
            i += 1
        else:
            sorted_list.append(right[j])
            j += 1

    # Add remaining elements
    sorted_list.extend(left[i:])
    sorted_list.extend(right[j:])

    return sorted_list


# Example dataset
numbers = [38, 27, 43, 3, 9, 82, 10]

# Display sorted output
print("Merge Sort Result:")
print(merge_sort(numbers))
```

```python
# Quick Sort Algorithm

def quick_sort(data):

    # Base condition
    if len(data) <= 1:
        return data

    # Select middle element as pivot
    pivot = data[len(data) // 2]

    # Store elements smaller than pivot
    left = [x for x in data if x < pivot]

    # Store elements equal to pivot
    middle = [x for x in data if x == pivot]

    # Store elements greater than pivot
    right = [x for x in data if x > pivot]

    # Recursively sort left and right lists
    return quick_sort(left) + middle + quick_sort(right)


# Example dataset
numbers = [10, 7, 8, 9, 1, 5]

# Display sorted output
print("Quick Sort Result:")
print(quick_sort(numbers))
```

```python
from collections import defaultdict, deque

# Graph class
class Graph:

    def __init__(self):
        self.graph = defaultdict(list)

    def add_edge(self, u, v, w=1):
        self.graph[u].append((v, w))
        self.graph[v].append((u, w))


# ---------------- CREATE GRAPH FIRST ----------------
g = Graph()

# Example edges
g.add_edge("A", "B", 4)
g.add_edge("A", "C", 2)
g.add_edge("B", "D", 5)
g.add_edge("C", "D", 1)


# ---------------- BFS FUNCTION ----------------
def bfs(graph, start):

    visited = set()
    queue = deque([start])
    result = []

    while queue:

        node = queue.popleft()

        if node not in visited:

            visited.add(node)
            result.append(node)

            for neighbour, _ in graph.graph[node]:
                queue.append(neighbour)

    return result


# ---------------- RUN BFS ----------------
print("BFS Traversal:", bfs(g, "A"))
```

```python
def dfs(graph, start, visited=None):

    if visited is None:
        visited = set()

    visited.add(start)
    result = [start]

    for neighbour, _ in graph.graph[start]:

        if neighbour not in visited:
            result.extend(dfs(graph, neighbour, visited))

    return result


# Example
print("DFS Traversal:", dfs(g, "A"))
```

```python
import heapq

def dijkstra(graph, start):

    distances = {node: float('inf') for node in graph.graph}
    distances[start] = 0

    pq = [(0, start)]

    while pq:

        current_dist, node = heapq.heappop(pq)

        for neighbour, weight in graph.graph[node]:

            distance = current_dist + weight

            if distance < distances[neighbour]:

                distances[neighbour] = distance
                heapq.heappush(pq, (distance, neighbour))

    return distances


# Example
print("Dijkstra Result:", dijkstra(g, "A"))
```

```python
# Stack implementation
class Stack:

    def __init__(self):
        self.items = []

    # push item into stack
    def push(self, item):
        self.items.append(item)

    # remove top item
    def pop(self):
        return self.items.pop()

    # view top element
    def peek(self):
        return self.items[-1]

    # check if stack is empty
    def isEmpty(self):
        return len(self.items) == 0


# Example: Parentheses matching
stack = Stack()

expression = "(A+B)"

for char in expression:

    if char == "(":
        stack.push(char)

    elif char == ")":
        stack.pop()

print("Balanced Expression:", stack.isEmpty())
```

```python
# Queue implementation
class Queue:

    def __init__(self):
        self.items = []

    # insert item
    def enqueue(self, item):
        self.items.append(item)

    # remove front item
    def dequeue(self):
        return self.items.pop(0)

    # display front item
    def front(self):
        return self.items[0]

    # check empty queue
    def isEmpty(self):
        return len(self.items) == 0


# Example: Task scheduling
q = Queue()

q.enqueue("Task 1")
q.enqueue("Task 2")

print("Current Task:", q.dequeue())
```

```python
# BST Node
class Node:

    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


# BST implementation
class BST:

    # insert node
    def insert(self, root, value):

        if root is None:
            return Node(value)

        if value < root.value:
            root.left = self.insert(root.left, value)

        else:
            root.right = self.insert(root.right, value)

        return root

    # inorder traversal
    def inorder(self, root):

        if root:
            self.inorder(root.left)
            print(root.value, end=" ")
            self.inorder(root.right)


# Example
tree = BST()

root = tree.insert(None, 50)
tree.insert(root, 30)
tree.insert(root, 70)

print("BST Traversal:")
tree.inorder(root)
```

```python
# Hash Table using chaining
class HashTable:

    def __init__(self, size):
        self.table = [[] for _ in range(size)]

    # hash function
    def hash_function(self, key):
        return hash(key) % len(self.table)

    # insert data
    def insert(self, key, value):

        index = self.hash_function(key)
        self.table[index].append((key, value))

    # search key
    def search(self, key):

        index = self.hash_function(key)

        for k, v in self.table[index]:

            if k == key:
                return v


# Example: Dictionary system
ht = HashTable(10)

ht.insert("name", "John")

print("Search Result:", ht.search("name"))
```
